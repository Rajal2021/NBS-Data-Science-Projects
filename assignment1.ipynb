# TASK:3 Hospital Temperature Checker

```python
# Takes the patient's name as input. 
patient_name = input("Patient's name:")
#Takes the patient's temperature in Celsius as input
temperature =float( input("Temperature in Celsius:"))
#Check temperature using if-elif-else
if temperature <= 37:
        category = "Normal"
elif 37<=temperature<=38.9:
        category= "Fever"
else:
        category = "High Fever"
#Step 4, 5 & 6: Display the results
print("Patient's Name:",patient_name)
print("Patient's Temperature:",temperature)
print("Category:",category)
```
